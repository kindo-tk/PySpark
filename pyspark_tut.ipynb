# PySpark Tutorial

```python
import pyspark
```

### Initiating a spark session

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
spark
```

### Reading the csv data using PySpark

```python
df_pyspark = spark.read.csv("Sample_data.csv", header=True, inferSchema=True)

## If inferSchema is mot used then by default all the columns are treated as string
```

### Showing the data read by PySpark

```python
df_pyspark.show()
```

```python
type(df_pyspark)
```

- .head() is used to show the content of the dataframe, if we don't provide anything inside head, by-default it shows the first row

```python
df_pyspark.head()
```

```python
df_pyspark.head(3)
```

- .printSchema() is tells the type of data present in a particular column, similar to .info() in pandas

```python
df_pyspark.printSchema()
```

### List all the columns

```python
df_pyspark.columns
```

### To list / show the contents of a particular column

```python
## Shows all the elements in the column country
df_pyspark.select('country').show()
```

```python
### Showing multiple columns
```

```python
df_pyspark.select(['account_number','country']).show()
```

### To check the datatype present in a particular column

```python
df_pyspark.dtypes
```

### Statistical overview, same as .describe() in pandas

```python
df_pyspark.describe().show()
```

### Adding columns in a dataframe

```python
df_pyspark = df_pyspark.withColumn('Amount_double',df_pyspark.amount*2)
```

```python
df_pyspark.show()
```

### Dropping the columns

```python
df_pyspark = df_pyspark.drop('Amount_double')
```

```python
df_pyspark.show()
```

### Renaming the column

```python
df_pyspark = df_pyspark.withColumnRenamed('transaction_type','Type_of_transaction')
```
