# PySpark Tutorial

```python
import pyspark
```

### Initiating a spark session

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
spark
```

### Reading the csv data using PySpark

```python
df_pyspark = spark.read.csv("Sample_data.csv", header=True)
```

### Showing the data read by PySpark

```python
df_pyspark.show()
```

```python
type(df_pyspark)
```
