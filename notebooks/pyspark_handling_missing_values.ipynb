## Handling Missing values using PySpark

```python
import pyspark
```

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('practise').getOrCreate()
```

### Reading the dataset

```python
df = spark.read.csv('../datasets/Sample_data_with_missing_values.csv', header  = True, inferSchema=True)
```

```python
df.show()
```

### Dropping a column

```python
df.drop('Name').show() 
## this does not change the original data frame we need to save in a variable in order to preserve the change
```

```python
df.show()
```

### Removing records containing the null values

```python
df.na.drop().show()
```

- drop(how = 'any') : drops the entire row if atleast any value in the row is null (default)
- drop(how = 'all') : drops the entire row if all values in the row is null

```python
df.na.drop(how = 'all').show()
```

- threshold defines that atleast that many non-null values must be present

```python
df.na.drop(how = 'all', thresh=2).show()

## all rows must have atleast 2 non-null values else the row will be deleted
```

```python
df.na.drop(how = 'all', thresh=3).show()
```

- Subset, is used to delete the entire row if that particular column has any null values

```python
df.na.drop(subset=['Experience']).show()
```

```python
df.na.drop(subset=['Name']).show()
```

### Filling the missing values

```python
df.na.fill('Missing Values').show() 

## Replaces all nan values with the text written here 'Missing Values' for columns having type string
```

```python
df.na.fill('Missing Values',['Experience','age']).show() 

## This is not filling as the datatype is different 
```

```python
df.na.fill(0,['Experience','age']).show() 
```

### Imputing the missing values with the mean or median value of that particular column

```python
from pyspark.ml.feature import Imputer
```

```python
imputer = Imputer(
    inputCols=['age','Experience','Salary'],
    outputCols=['{}_imputed'.format(c) for c in ['age','Experience','Salary']]
).setStrategy('mean')
```

```python
imputer.fit(df).transform(df).show()
```

```python
imputer = Imputer(
    inputCols=['age','Experience','Salary'],
    outputCols=['{}_imputed'.format(c) for c in ['age','Experience','Salary']]
).setStrategy('median')

imputer.fit(df).transform(df).show()
```

```python
from pyspark.sql import SparkSession

from pyspark.sql.functions import (
    col, count, when, lit, isnan, sum, round, avg, mean, coalesce, isnull
)
```

```python
spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
df = spark.read.csv('../datasets/transactions_missing.csv', header = True, inferSchema=True)

df.show(truncate = False)
```

```python
df.printSchema()
```

## checking null values in each column

```python
null_counts = df.select([
    count(when(col(c).isNull(),1)).alias(c)
    for c in df.columns
])

null_counts.show(truncate=False)
```

```python
# Null percentage 
total_rows = df.count()

null_pct = df.select([
    round(
        count(when(col(c).isNull(), 1)) / total_rows * 100,
        2
    ).alias(c)
    for c in df.columns
])

null_pct.show(truncate=False)
```

`In PySpark there are two kinds of "missing":`

- null — SQL null, represents absence of a value. Comes from missing CSV fields, failed joins, explicit None.
- NaN — "Not a Number", a special float value. Comes from 0/0 or explicit float('nan'). Only applies to float/double columns.

```python
from pyspark.sql.functions import isnan
```

```python
# Check for BOTH null AND NaN in numeric columns
df.select([
    count(
        when(col(c).isNull() | isnan(col(c)), 1)
    ).alias(c)
    for c in ["balance", "transaction_amount", "credit_score", "annual_income"]
]).show()
```

## Dropping records containing null values

```python
df_dropped = df.na.drop()
# Equivalent:
df_dropped = df.dropna()
```

```python
print(f"Before: {df.count()} rows")
print(f"After:  {df_dropped.count()} rows")
```

```python
df_dropped_all = df.na.drop(how="all")
# drops a record if every column in that record is null
# default is how = 'any'
```

## Filling missing values in specific columns 

```python
# Fill only specific columns
df_filled = df.na.fill({
    "account_type":           "Unknown",
    "relationship_manager":   "Unassigned",
    "num_previous_defaults":  0,
    "city":                   "Unknown"
})

df_filled.show()
```

## Fill missing values using another column's values

```python
from pyspark.sql.functions import coalesce
```

```python
# If credit_score is null, use a fallback score derived from defaults
# If that's also null, default to 300 (worst score)
df = df.withColumn("credit_score_clean",
    coalesce(
        col("credit_score"),
        when(col("num_previous_defaults") == 0, lit(650)),
        lit(300)
    )
)
```

coalesce(col1, col2, col3) — tries col1 first. If null, tries col2. If null, tries col3. Returns the first non-null.

### Detecting duplicate rows

```python
total   = df.count()
unique  = df.distinct().count()
duplicates = total - unique

print(f"Total rows:      {total}")
print(f"Unique rows:     {unique}")
print(f"Duplicate rows:  {duplicates}")
```

```python
# Deduplicate based on specific columns only
# Keep only one transaction per customer per date
df_deduped = df.dropDuplicates(["customer_id", "transaction_date"])
```
