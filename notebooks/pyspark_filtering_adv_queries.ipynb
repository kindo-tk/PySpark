# Filtering commands and advanced queries in PySpark

### Importing the necessary libraries

```python
import pyspark
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType, DateType
```

```python
spark = SparkSession.builder.appName("Practise").getOrCreate()
```

```python
spark
```

### Defining the schema of the data

```python
schema = StructType([
    StructField("transaction_id",     StringType(),  True),
    StructField("customer_id",        StringType(),  True),
    StructField("name",               StringType(),  True),
    StructField("age",                IntegerType(), True),
    StructField("account_type",       StringType(),  True),
    StructField("balance",            FloatType(),   True),
    StructField("transaction_amount", FloatType(),   True),
    StructField("transaction_type",   StringType(),  True),
    StructField("city",               StringType(),  True),
    StructField("transaction_date",   StringType(),  True),
    StructField("credit_score",       IntegerType(), True),
])
```

### Reading the data

```python
df = spark.read.csv('../datasets/transactions.csv', header = True, schema = schema)
```

### Show data where balance is greater than 50000

```python
df.filter("balance>50000").show()
```

```python
df.where("balance>50000").show()
```

- where and filter are 100% identical and can be used interchangeably

- we also have column expression which gives more flexibility

```python
df.filter(col('balance')>50000).show()
```

### Single condition filters

```python
df.filter(col('transaction_type')=='Debit').show()
```

```python
df.filter(col('transaction_amount')==5000).show()
```

```python
df.filter(col('account_type')!='Savings').show()
```

### Filter with multiple conditions using (and, or)

```python
df.filter(
    (col('transaction_type')=='Debit') &
    (col('transaction_amount')>3000)
).show()
```

```python
df.filter(
    (col("city") == "Mumbai") | (col("city") == "Delhi")
).show()
```

```python
df.filter(~(col("transaction_type") == "Credit")).show()
```

### Range filter using between

- the upper and lower bounds are inclusive

```python
df.filter(col('credit_score').between(600,700)).show()
```

### List filter with isin()

```python
metro_cities = ['Mumbai','Delhi','Bangalore','Kolkata','Chennai']

df.filter(col('city').isin(metro_cities)).show()
```

```python
df.filter(~col('account_type').isin(['Savings','Current'])).show()
```

### String filters


```python
from pyspark.sql.functions import col, lower, upper
```

- names starting with A

```python
df.filter(col('name').like('A%')).show()
```

- this pattern matching is similar to sql and it is case sensitive, meaning A and a are different 

```python
# Case-insensitive: names containing 'kumar' in any case
df.filter(lower(col("name")).contains("sharma")).show()
```

```python
# Starts with / ends with
df.filter(col("name").startswith("Pr")).show()
df.filter(col("name").endswith("Sen")).show()
```

### Null filters

```python
## returns all the rows where credit_score is null
df.filter(col("credit_score").isNull()).show()
```

- similarly we have a function isNotNULL that returns rows containing non-null values

```python
df.filter(col('credit_score').isNotNull()).show()
```

### Chaining multiple conditions

```python
## high-value debits from savings, credit score below 650
result = df \
    .filter(col("transaction_type") == "Debit") \
    .filter(col("account_type") == "Savings") \
    .filter(col("transaction_amount") > 3000) \
    .filter(col("credit_score") < 650) \
    .filter(col("credit_score").isNotNull())

result.show()
```

### Filtering and then selecting only the required columns

```python
result = df \
    .filter(col("transaction_type") == "Debit") \
    .filter(col("balance") < 20000) \
    .select("customer_id", "name", "balance", "transaction_amount", "city")

result.show()
```
