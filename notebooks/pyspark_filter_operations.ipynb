# Filter Operations Using PySpark

```python
import pyspark
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Practise').getOrCreate()
```

```python
df = spark.read.csv("../datasets/sample_data_3.csv", header=True, inferSchema=True)
```

```python
df.show()
```

## Filter operations

- salary of persons less than or equal to 20000

```python
df.filter("Salary<=20000").show()
```

## Alternative syntax

```python
df.filter(df['Salary']<=20000).show()
```

- selecting specific columns after filtration

```python
df.filter("Salary<=20000").select(['Name','Salary']).show()
```

### Using multiple conditions together

```python
df.filter(
    (df['Salary']<=20000) &
    (df['Salary']>=15000)
).show()
```

### Inverse filter operation

```python
df.filter(~(df['Salary']<=20000)).show()
```
